```python
## Grid Independence Study ##
coarse= pd.read_csv('/content/drive/MyDrive/CD_Nozzle/isentropic_26/gis/co_avg.csv')
medium= pd.read_csv('/content/drive/MyDrive/CD_Nozzle/isentropic_26/gis/me_avg.csv')
fine= pd.read_csv('/content/drive/MyDrive/CD_Nozzle/isentropic_26/gis/fi_avg.csv')
finer= pd.read_csv('/content/drive/MyDrive/CD_Nozzle/isentropic_26/gis/ffi_avg.csv')
an_values = pd.read_csv('/content/drive/MyDrive/CD_Nozzle/isentropic_26/analytical.csv')

#coarse.columns
plt.plot(coarse["Points:0"], coarse["Pressure"], "b")
plt.plot(medium["Points:0"], medium["Pressure"], "g")
plt.plot(fine["Points:0"], fine["Pressure"], "r")
plt.plot(finer["Points:0"], finer["Pressure"], "y")
plt.plot(an_values["x_values"], an_values["P_an26"], color="magenta")
plt.grid(True)
plt.xlabel("X Position (m)")
plt.ylabel("Pressure (Pa)")
plt.legend(["Coarse", "Medium", "Fine", "Finer", "Analytical"])
plt.title("Grid Independence Study")
plt.show()
```
