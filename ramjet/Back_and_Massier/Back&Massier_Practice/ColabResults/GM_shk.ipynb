```python
plt.plot(bac["X_norm"], bac["P_norm"])
plt.plot(bac["X_exnorm"], bac["P_exnorm"], "^")
plt.xlabel("X/L")
plt.ylabel("Normalized Pressure")
plt.legend(("SU2", "Back et al. Exp"))
plt.show()
```

```python
# Variation of mach number along symmetry line for different mesh size
#plt.plot(coarse["Points:0"], coarse["Mach"], "r")
plt.plot(med["Points:0"], med["Mach"])
plt.plot(fine["Points:0"], fine["Mach"], "g")
plt.plot(finer["Points:0"], finer["Mach"], "y")
plt.plot(finest["Points:0"], finest["Mach"], "m")
plt.xlabel("X Position (m)")
plt.ylabel("Mach Number")
plt.legend(("Coarse", "Medium", "Fine", "Finer", "Finest"))
plt.show()
```
