```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# LOAD FILE
# =====================================================

file_path = r"C:\Users\VEDANT\Desktop\ramjet\extend 1 m, no slip\flow.csv"

df = pd.read_csv(file_path)

# =====================================================
# EXTRACT x = 1 m PLANE
# =====================================================

x_target = 1.0

# Increase tolerance if no points found

tol = 1e-3

section = df[
    np.abs(df['Points_0'] - x_target) < tol
].copy()

# =====================================================
# SORT BY y
# =====================================================

section = section.sort_values(by='Points_1')

# =====================================================
# SAVE DATA
# =====================================================

section.to_csv(
    r"C:\Users\VEDANT\Desktop\velocity_profile_x1m_noslip.csv",
    index=False
)

# =====================================================
# PLOT
# =====================================================

plt.figure(figsize=(8,6))

plt.plot(
    section['Velocity_Magnitude'],
    section['Points_1'],
    linewidth=3
)

plt.xlabel(
    'Velocity Magnitude (m/s)',
    fontsize=16
)

plt.ylabel(
    'y Coordinate (m)',
    fontsize=16
)

plt.title(
    'Velocity Profile at x = 1.0 m',
    fontsize=16
)

plt.grid(True)

plt.tight_layout()

plt.savefig(
    r"C:\Users\VEDANT\Desktop\velocity_profile_x1m.png",
    dpi=600
)

plt.show()

print("Number of points found =", len(section))
```
