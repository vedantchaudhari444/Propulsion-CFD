```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Read the CSV file
df = pd.read_csv("/content/drive/MyDrive/Chen/x1.csv")  # Replace with your actual filename

# Step 2: Extract time and pressure
time = df['timec']
pressure = df['avg(Pconv)']

# Step 3: Compute fluctuating component (p′)
mean_pressure = pressure.mean()
p_prime = pressure - mean_pressure

#
time = time*10**-5

# Step 4: Plot p′ vs time
plt.figure(figsize=(10, 5))
plt.plot(time, p_prime, label="p′ (fluctuating pressure)", color='blue')
plt.xlabel("Time")
plt.ylabel("p′ (Pa)")
plt.title("Fluctuating Pressure vs Time")

# Step 5: Add scale bar representing 17.24 kPa
# Choose position near the end of the time axis
x_pos = time.iloc[-1] * 0.98
y_start = p_prime.min() + 0.1 * (p_prime.max() - p_prime.min())
y_end = y_start + (p_prime.max() - p_prime.min()) * 0.3  # adjust for scale height

# Draw the vertical bar and label
plt.vlines(x=x_pos, ymin=y_start, ymax=y_end, color='black', linewidth=2)
plt.text(x_pos + (time.max() - time.min()) * 0.01, (y_start + y_end) / 2,
         "17.24 kPa", va='center', ha='left', fontsize=10, color='black')

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print(mean_pressure)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Read the CSV file
# Make sure the 'x12.csv' file is accessible in your Colab environment.
# If it's in Google Drive, you'll need to mount your Drive first:
# from google.colab import drive
# drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Chen/x12.csv') # Example path if in Drive
#df = pd.read_csv('x12.csv') # Use this if x12.csv is in the same directory as your notebook

# Step 2: Extract time and pressure
time = df['Timec']
pressure = df['Pec']

# Step 3: Compute fluctuating component (p′)
# Drop NaN values from pressure before computing mean to avoid issues
pressure_cleaned = pressure.dropna()
mean_pressure = pressure_cleaned.mean()
p_prime = pressure - mean_pressure

# Step 4: Apply a moving average for smoothing
# Choose a window size for the moving average.
# A smaller window gives less smoothing, a larger window gives more.
# You can experiment with this value (e.g., 3, 7, 10) to find the desired smoothness.
window_size = 5
p_prime_smoothed = p_prime.rolling(window=window_size, min_periods=1, center=True).mean()

# Step 5: Adjust time scale as per your original request
time = time * 10**-5

# Step 6: Plot both original and smoothed p′ vs time
plt.figure(figsize=(12, 6))

# Plot the original data with steps (optional, but good for comparison)
plt.plot(time, p_prime, label="Original p′ (fluctuating pressure)", color='lightgray', linestyle='--', alpha=0.7)

# Plot the smoothed data
plt.plot(time, p_prime_smoothed, label=f"Smoothed p′ (Moving Average, Window={window_size})", color='blue')

plt.xlabel("Time")
plt.ylabel("p′ (Pa)")
plt.title("Fluctuating Pressure vs Time (Original and Smoothed)")

# Step 7: Add scale bar representing 17.24 kPa
# Ensure there's data before attempting to draw the scale bar
if not p_prime_smoothed.empty and not time.empty:
    x_pos = time.iloc[-1] * 0.98 # Position the scale bar near the end of the x-axis

    y_min_smoothed = p_prime_smoothed.min()
    y_max_smoothed = p_prime_smoothed.max()
    y_range_smoothed = y_max_smoothed - y_min_smoothed

    # Calculate y_start and y_end for the scale bar, converting kPa to Pa (17.24 kPa = 17240 Pa)
    # The scale bar height is 17240 Pa. We position it near the bottom of the plot.
    scale_bar_height_pa = 17240
    y_start = y_min_smoothed + 0.1 * y_range_smoothed
    y_end = y_start + scale_bar_height_pa

    # Adjust y_start if y_end exceeds the plot limits too much, to keep the bar visible
    if y_end > y_max_smoothed * 1.1: # if it goes significantly above the max y, adjust downwards
        y_end = y_max_smoothed * 1.1
        y_start = y_end - scale_bar_height_pa

    plt.vlines(x=x_pos, ymin=y_start, ymax=y_end, color='black', linewidth=2)
    plt.text(x_pos + (time.max() - time.min()) * 0.01, (y_start + y_end) / 2,
             "17.24 kPa", va='center', ha='left', fontsize=10, color='black')

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show() # This command will display the plot in Colab
```
